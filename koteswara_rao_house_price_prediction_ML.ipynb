```python
df_classification = df.copy()

median_price = df_classification['Price'].median()

df_classification['Price_Category'] = (
    df_classification['Price'] >= median_price
).astype(int)

df_classification[['Price', 'Price_Category']].head()
```

Logistic Regression

```python
X_classification = df_classification.drop(
    columns=['Price', 'Price_Category']
)

y_classification = df_classification['Price_Category']
X_classification.head()
```

```python
y_classification.value_counts()
```

```python
X_train_classification, X_test_classification, y_train_classification, y_test_classification = train_test_split(
    X_classification,
    y_classification,
    test_size=0.20,
    train_size=0.80,
    random_state=32,
    stratify=y_classification
)
```

```python
X_train_classification = pd.get_dummies(
    X_train_classification,
    columns=['Location', 'Condition', 'Garage'],
    drop_first=True,
    dtype=int
)

X_test_classification = pd.get_dummies(
    X_test_classification,
    columns=['Location', 'Condition', 'Garage'],
    drop_first=True,
    dtype=int
)
```

```python
X_train_classification, X_test_classification = X_train_classification.align(
    X_test_classification,
    join='left',
    axis=1,
    fill_value=0
)
```

```python
scaler_classification = StandardScaler()

X_train_classification_scaled = scaler_classification.fit_transform(
    X_train_classification
)

X_test_classification_scaled = scaler_classification.transform(
    X_test_classification
)
```

```python
logistic_model = LogisticRegression(
    max_iter=1000
)
```

```python
logistic_model.fit(
    X_train_classification_scaled,
    y_train_classification
)

print("Logistic Regression model trained successfully!")
```

```python
y_pred_logistic = logistic_model.predict(
    X_test_classification_scaled
)

print("Logistic Regression predictions generated successfully!")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import roc_curve, roc_auc_score
```

```python
accuracy_logistic = accuracy_score(
    y_test_classification,
    y_pred_logistic
)

precision_logistic = precision_score(
    y_test_classification,
    y_pred_logistic
)

recall_logistic = recall_score(
    y_test_classification,
    y_pred_logistic
)

f1_logistic = f1_score(
    y_test_classification,
    y_pred_logistic
)

print("Logistic Regression Performance")
print("--------------------------------")

print("Accuracy:", accuracy_logistic)
print("Accuracy Percentage:", accuracy_logistic * 100, "%")

print("Precision:", precision_logistic)
print("Precision Percentage:", precision_logistic * 100, "%")

print("Recall:", recall_logistic)
print("Recall Percentage:", recall_logistic * 100, "%")

print("F1 Score:", f1_logistic)
print("F1 Score Percentage:", f1_logistic * 100, "%")
```

```python
cm_logistic = confusion_matrix(
    y_test_classification,
    y_pred_logistic
)

cm_logistic
```

```python
print(
    classification_report(
        y_test_classification,
        y_pred_logistic
    )
)
```

```python
y_prob_logistic = logistic_model.predict_proba(
    X_test_classification_scaled
)[:, 1]

fpr_logistic, tpr_logistic, thresholds_logistic = roc_curve(
    y_test_classification,
    y_prob_logistic
)

auc_logistic = roc_auc_score(
    y_test_classification,
    y_prob_logistic
)

print("Logistic Regression ROC-AUC")
print("---------------------------")
print("AUC:", auc_logistic)
```
